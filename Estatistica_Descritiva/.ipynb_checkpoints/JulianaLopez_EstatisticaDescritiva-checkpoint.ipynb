#### Medidas de dispersão
<br>Variância, desvios amostrais, amplitude, intervalo interquartílico, coeﬁciente de dispersão quartílica e coeﬁciente de variação
Interpretação
1. Sua amostra é concentrada ou dispersa?
2. A amplitude interna (IIQ) é proporcional à amplitude total da amostra?
3. Quais a implicações dessas informações em relação ao contexto/signiﬁcado concreto dessa variável?

```python
# Base completa
# variância
S2_Desmat = df_baseED["%Desmat"].var()
# desvio padrão 
S_Desmat = df_baseED["%Desmat"].std()
# desvios absoluto médio
DAM_Desmat = np.abs(df_baseED["%Desmat"] - df_baseED["%Desmat"].mean()).mean()
# desvios absoluto mediano
MAD_Desmat = np.median(np.absolute(df_baseED["%Desmat"] - df_baseED["%Desmat"].median()))
# amplitude
A_Desmat = df_baseED["%Desmat"].max() - df_baseED["%Desmat"].min()
print(A_Desmat)
# intervalo interquartílico
Q1_Desmat = df_baseED["%Desmat"].quantile(0.25)
Q3_Desmat = df_baseED["%Desmat"].quantile(0.75)
IIQ_Desmat = Q3_Desmat - Q1_Desmat
# coeﬁciente de dispersão quartílica 
CDQ_Desmat = IIQ_Desmat / (Q3_Desmat + Q1_Desmat)
# coeﬁciente de variação
CV_Desmat = (S_Desmat / df_baseED["%Desmat"].mean()) * 100
```

```python
# Base com filtro 0 e outliers
# variância
S2_Desmat_filtrado = df_baseED_filtrado["%Desmat"].var()
# desvio padrão 
S_Desmat_filtrado = df_baseED_filtrado["%Desmat"].std()
# desvios absoluto médio
DAM_Desmat_filtrado = np.abs(df_baseED_filtrado["%Desmat"] - df_baseED_filtrado["%Desmat"].mean()).mean()
# desvios absoluto mediano
MAD_Desmat_filtrado = np.median(np.absolute(df_baseED_filtrado["%Desmat"] - df_baseED_filtrado["%Desmat"].median()))
# amplitude
A_Desmat_filtrado = df_baseED_filtrado["%Desmat"].max() - df_baseED_filtrado["%Desmat"].min()
# intervalo interquartílico
Q1_Desmat_filtrado = df_baseED_filtrado["%Desmat"].quantile(0.25)
Q3_Desmat_filtrado = df_baseED_filtrado["%Desmat"].quantile(0.75)
IIQ_Desmat_filtrado = Q3_Desmat_filtrado - Q1_Desmat_filtrado
# coeﬁciente de dispersão quartílica 
CDQ_Desmat_filtrado = IIQ_Desmat_filtrado / (Q3_Desmat_filtrado + Q1_Desmat_filtrado)
# coeﬁciente de variação
CV_Desmat_filtrado = (S_Desmat_filtrado / df_baseED_filtrado["%Desmat"].mean()) * 100
```

#### Gráﬁco de dispersão (KDE)
1. Sinalize os componentes do IIQ (i.e. Q1 e Q3) com linhas verticais na cor cinza;

```python
ax = sns.displot(df_baseED, x='%Desmat', stat='density', kde=True)
ax.set(xlabel='Variação', ylabel='Densidade')

# IIQ
plt.axvline(x=Q1_Desmat, color='#a0a0a0ff')
plt.axvline(x=Q3_Desmat, color='#a0a0a0ff')

# Desvio Padrao
xbar_Desmat = df_baseED["%Desmat"].mean()
plt.axvline(x=xbar_Desmat, color='#a0a0c0ff')
plt.axvline(x=xbar_Desmat-S_Desmat, color='#a0a0c0ff')
plt.axvline(x=xbar_Desmat+S_Desmat, color='#a0a0c0ff')
plt.yscale("log") 
```

#### Medidas de associação: duas variáveis categóricas


1. V de Cramer

```python

# Cria a tabela de contingência
tabela = pd.crosstab(df_baseED['Categoria (Sigla)'], df_baseED['Bioma'])
display(tabela)

# Aplica o teste qui-quadrado
chi2, p, dof, expected = chi2_contingency(tabela)

# Número total de observações
n = tabela.values.sum()
display(tabela.shape)
# Número mínimo de categorias menos 1 (para linhas ou colunas)
min_dim = min(tabela.shape) - 1
print(min_dim)
# Cálculo do V de Cramer
cramers_v = np.sqrt(chi2 / (n * min_dim))
print("V de Cramer:", cramers_v)
```

2. Qui-Quadrado

```python
chi2_val, p_val, dof_val, expected_val = chi2_contingency(tabela)
print("Teste Qui-Quadrado:")
print("  Qui-quadrado:", chi2_val)
print("  p-valor:", p_val)
print("  Graus de Liberdade:", dof_val)
```

3. Coeficiente de contingência

```python
contingency_coefficient = np.sqrt(chi2 / (chi2 + n))
print("Coeficiente de Contingência:", contingency_coefficient)
```

## Regressão linear simples (duas variáveis contínuas)

```python
x=df_baseED['área_ha_uc']  # Variável no eixo x
y=df_baseED['TotalDesmat']    # Variável no eixo y
data=df_baseED     # DataFrame
X = sm.add_constant(x)
```

1. OLS

```python
# Estimação por OLS (que é MLE sob erro normal)
model = sm.OLS(y, X).fit()
print(model.summary())
print("\n")

# Fazer previsões
y_pred = model.predict(X)

# Visualizar resultados
plt.scatter(x, y, label='Dados reais')
plt.plot(x, y_pred, color='red', label='Regressão Linear')
plt.xlabel('Variável Independente (Area UC)')
plt.ylabel('Variável Dependente (% Área desmatada)')
plt.xscale('log')
plt.yscale('log')
plt.legend()
plt.show()
```

2. MLE

```python
import scipy.optimize as opt

# Definindo a função de log-verossimilhança negativa
# Usamos a parametrização: params = [beta0, beta1, log_sigma]
def neg_log_likelihood(params, x, y):
    beta0, beta1, log_sigma = params
    sigma = np.exp(log_sigma)  # Garante que sigma > 0
    n = len(y)
    residuals = y - (beta0 + beta1 * x)
    # Função de log-verossimilhança negativa:
    nll = 0.5 * n * np.log(2 * np.pi) + n * log_sigma + np.sum(residuals**2) / (2 * sigma**2)
    return nll

# Chute inicial para [beta0, beta1, log_sigma]
initial_guess = [0, 0, 0]

# Minimiza a log-verossimilhança negativa
result = opt.minimize(neg_log_likelihood, initial_guess, args=(x, y))

beta0_hat, beta1_hat, log_sigma_hat = result.x
sigma_hat = np.exp(log_sigma_hat)

print("Estimativas MLE:")
print("beta0 =", beta0_hat)
print("beta1 =", beta1_hat)
print("sigma =", sigma_hat)

# Visualização dos dados e da reta ajustada
plt.scatter(x, y, label='Dados')
plt.plot(x, beta0_hat + beta1_hat * x, color='red', label='Reta MLE')
plt.xlabel('Variável Independente (Area UC)')
plt.ylabel('Variável Dependente (% Área desmatada)')
plt.xscale('log')
plt.legend()
plt.title("Regressão Linear via MLE")
plt.show()
```

3. M-estimador de Huber

```python
from sklearn.linear_model import HuberRegressor

# Regressão robusta usando Huber
huber = HuberRegressor(epsilon=1.35)
huber.fit(X, y)

# Fazer previsões
y_pred = huber.predict(X)

# Parâmetros estimados
print(f"Coeficiente: {huber.coef_}")
print(f"Intercepto: {huber.intercept_}")

# Visualizar resultados
plt.scatter(x, y, label='Dados reais')
plt.plot(x, y_pred, color='red', label='Regressão Linear')
plt.xlabel('Variável Independente (Area UC)')
plt.ylabel('Variável Dependente (% Área desmatada)')
plt.xscale('log')
plt.legend()
plt.show()
```

4. Quantílico

#### Decomponha e analise os componentes da série temporal (tendência, sazonalidade e ruído).
- Tendência: Refere-se à direção de longo prazo dos dados. Pode ser ascendente, descendente ou constante. Ex: Aumento contínuo do Produto Interno Bruto (PIB) ao longo dos anos.
- Sazonalidade: São padrões periódicos e previsíveis que ocorrem em intervalos regulares (por exemplo, variações mensais ou sazonais). Ex: Aumento nas vendas de determinado produto durante as festas de ﬁnal de ano.
- Ruído/Resíduo: Refere-se a ﬂutuações irregulares, que podem representar erros pontuais de coleta ou até mesmo choques externos (e.g. fatores econômicos ou políticos). Ex: Guerras, Ciclos econômicos de expansão e recessão.
