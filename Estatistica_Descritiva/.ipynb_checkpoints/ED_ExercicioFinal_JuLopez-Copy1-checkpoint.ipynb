### Medidas de dispersão: variância, desvios amostrais, amplitude, intervalo interquartílico, coeﬁciente de dispersão quartílica e coeﬁciente de variação
Interpretação
1. Sua amostra é concentrada ou dispersa?
2. A amplitude interna (IIQ) é proporcional à amplitude total da amostra?
3. Quais a implicações dessas informações em relação ao contexto/signiﬁcado concreto dessa variável?

```python
# Base completa
# variância
S2_Desmat = df_baseED["%Desmat"].var()
print(S2_Desmat)

# desvio padrão 
S_Desmat = df_baseED["%Desmat"].std()
print(S_Desmat)

# desvios absoluto médio
DAM_Desmat = np.abs(df_baseED["%Desmat"] - df_baseED["%Desmat"].mean()).mean()
print(DAM_Desmat)

# desvios absoluto mediano
MAD_Desmat = np.median(np.absolute(df_baseED["%Desmat"] - df_baseED["%Desmat"].median()))
print(MAD_Desmat)

# amplitude
A_Desmat = df_baseED["%Desmat"].max() - df_baseED["%Desmat"].min()
print(A_Desmat)

# intervalo interquartílico
Q1_Desmat = df_baseED["%Desmat"].quantile(0.25)
Q3_Desmat = df_baseED["%Desmat"].quantile(0.75)

IIQ_Desmat = Q3_Desmat - Q1_Desmat
print(IIQ_Desmat)

# coeﬁciente de dispersão quartílica 
CDQ_Desmat = IIQ_Desmat / (Q3_Desmat + Q1_Desmat)
print(CDQ_Desmat)

# coeﬁciente de variação
CV_Desmat = (S_Desmat / df_baseED["%Desmat"].mean()) * 100
print(CV_Desmat)
```

```python
# Base completa filtro 0 e outliers
# variância
S2_Desmat_filtrado = df_baseED_filtrado["%Desmat"].var()
print(S2_Desmat_filtrado)

# desvio padrão 
S_Desmat_filtrado = df_baseED_filtrado["%Desmat"].std()
print(S_Desmat_filtrado)

# desvios absoluto médio
DAM_Desmat_filtrado = np.abs(df_baseED_filtrado["%Desmat"] - df_baseED_filtrado["%Desmat"].mean()).mean()
print(DAM_Desmat_filtrado)

# desvios absoluto mediano
MAD_Desmat_filtrado = np.median(np.absolute(df_baseED_filtrado["%Desmat"] - df_baseED_filtrado["%Desmat"].median()))
print(MAD_Desmat_filtrado)

# amplitude
A_Desmat_filtrado = df_baseED_filtrado["%Desmat"].max() - df_baseED_filtrado["%Desmat"].min()
print(A_Desmat_filtrado)

# intervalo interquartílico
Q1_Desmat_filtrado = df_baseED_filtrado["%Desmat"].quantile(0.25)
Q3_Desmat_filtrado = df_baseED_filtrado["%Desmat"].quantile(0.75)

IIQ_Desmat_filtrado = Q3_Desmat_filtrado - Q1_Desmat_filtrado
print(IIQ_Desmat_filtrado)

# coeﬁciente de dispersão quartílica 
CDQ_Desmat_filtrado = IIQ_Desmat_filtrado / (Q3_Desmat_filtrado + Q1_Desmat_filtrado)
print(CDQ_Desmat_filtrado)

# coeﬁciente de variação
CV_Desmat_filtrado = (S_Desmat_filtrado / df_baseED_filtrado["%Desmat"].mean()) * 100
print(CV_Desmat_filtrado)
```

## Assimetria e curtose
#### Assimetria
Assimetria Positiva: Cauda à direita longa ou pesada; Média maior que mediana, que é maior que a moda

Assimetria Negativa: Cauda longa à esquerda; Moda maior que mediana, que é maior que a média

Distribuição Simétrica: Média, mediana e moda são coincidentes; Forma de sino com dados distribuídos igualmente à direita e à esquerda

Terceiro Momento Central: Medida robusta de assimetria; Dá a intensidade e direção da assimetria

Coeﬁciente de Fisher-Pearson: Medida ajustada a partir do terceiro momento central. Tem a mesma interpretação, mas tende a corrigir o viés de amostras menores. Mais comumente utilizado entre os coeﬁcientes de assimetria. 

Primeiro Coeﬁcientes de Pearson:
1. Compara a média com a moda. Em uma distribuição simétrica, média e moda são iguais. 
2. Se a média amostral for maior que a moda, haverá assimetria positiva; se for menor, negativa.
3. A diﬁculdade está em deﬁnir a moda de forma consistente, especialmente em distribuições contínuas.

Segundo Coeﬁciente de Pearson:
1. Multiplica por 3 a diferença entre a média e a mediana, padronizada pelo desvio padrão. 
2. Assume que a mediana é uma boa aproximação para a moda em distribuições levemente assimétricas e é mais prática de calcular. 
3. É amplamente utilizada como um indicador rápido da direção da assimetria

Coeﬁciente Quartílico de Bowle
1. Medida bastante simples e robusta. 
2. Menos sensível a outliers. 
3. É muito útil para obter uma visão rápida sobre a simetria dos dados no centro da amostra

#### Curtose
A curtose mede o "peso" das caudas de uma distribuição e a concentração dos dados em torno da média, ou seja, o quão pontiaguda" ou "achatada" a distribuição é em comparação com uma distribuição normal.
❏ Uma distribuição com curtose igual à da normal (geralmente 3 quando usamos a deﬁnição tradicional) é considerada mesocúrtica.
❏ Se a curtose for maior que 3, a distribuição é leptocúrtica (mais pontiaguda e com caudas mais pesadas).
❏ Se a curtose for menor que 3, a distribuição é platicúrtica (mais achatada e com caudas mais leves).

Quarto momento amostral: Calculado de forma semelhante à assimetria

Excesso de curtose: Como a distribuição normal possui uma curtose igual a 3, é comum calcular o excesso de curtose, que é a diferença entre a curtose da amostra e 3:
❏ Excesso positivo: indica caudas mais pesadas do que as da normal (distribuição leptocúrtica).
❏ Excesso negativo: indica caudas mais leves (distribuição platicúrtica)

Detecção de Valores Extremos:
❏ Auxilia na identiﬁcação de outliers e valores atípicos no conjunto de dados
❏ Uma alta curtose indica maior probabilidade de valores extremos na distribuição 
❏ Distribuições com curtose elevada sugerem maior risco de eventos extremos

Normalidade dos Dados:
❏ Auxilia na avaliação da normalidade quando combinada com outras métricas estatísticas
❏ Inﬂuencia a escolha de métodos estatísticos apropriados para análise
❏ Pode indicar necessidade de normalização dos dados através de transformações especíﬁcas

## Medidas de associação: duas variáveis categóricas

### V de Cramer

```python
# Cria a tabela de contingência
tabela = pd.crosstab(df_baseED['faixas'], df_baseED['Bioma'])
display(tabela)

# Aplica o teste qui-quadrado
chi2, p, dof, expected = chi2_contingency(tabela)

# Número total de observações
n = tabela.values.sum()
display(tabela.shape)
# Número mínimo de categorias menos 1 (para linhas ou colunas)
min_dim = min(tabela.shape) - 1
print(min_dim)
# Cálculo do V de Cramer
cramers_v = np.sqrt(chi2 / (n * min_dim))
print("V de Cramer:", cramers_v)

```

## Regressão linear simples

### Coeficiente de Determinação R²

```python
from sklearn.metrics import r2_score

# Dados simulados
x_reg = df_baseED['área_ha_uc']
y_reg = 2 * x_reg + np.random.normal(0, 1, len(x_reg))  # y = 2x + ruído

# Ajuste de uma reta (regressão linear) usando polyfit
coeffs = np.polyfit(x_reg, y_reg, 1)
y_pred = np.polyval(coeffs, x_reg)

# Cálculo manual do R²
ss_res = np.sum((y_reg - y_pred) ** 2)
ss_tot = np.sum((y_reg - np.mean(y_reg)) ** 2)
r2_manual = 1 - ss_res / ss_tot

# Ou usando r2_score do sklearn
r2_sklearn = r2_score(y_reg, y_pred)

print("Coeficiente de Determinação (R²) - Manual:", r2_manual)
print("Coeficiente de Determinação (R²) - sklearn:", r2_sklearn)
```
