```python
df_desmatAm = pd.read_excel('https://github.com/JuPLopez/JL_mba_enap/raw/refs/heads/main/Estatistica_Descritiva/tabela_de_desmatamento_ucs_atualizado_prodes-2023(1).xlsx', skiprows= 1, skipfooter=1)
df_desmatAm.drop(['TOTAL', 'Até 2007'], axis=1, inplace=True)
df_desmatMA = pd.read_excel('https://github.com/JuPLopez/JL_mba_enap/raw/refs/heads/main/Estatistica_Descritiva/tabela_de_desmatamento_ucs_atualizado_prodes-2023(1).xlsx', skiprows= 1, skipfooter=1, sheet_name = 'Mata Atlântica')
df_desmatMA.drop(['TOTAL', 'Até 2000', '2001 a 2004',2006], axis=1, inplace=True)
df_desmatPp = pd.read_excel('https://github.com/JuPLopez/JL_mba_enap/raw/refs/heads/main/Estatistica_Descritiva/tabela_de_desmatamento_ucs_atualizado_prodes-2023(1).xlsx', skiprows= 1, skipfooter=1, sheet_name = 'Pampa')
df_desmatPp.drop(['TOTAL', 'Até 2000', '2001 a 2004',2006], axis=1, inplace=True)
df_desmatCe = pd.read_excel('https://github.com/JuPLopez/JL_mba_enap/raw/refs/heads/main/Estatistica_Descritiva/tabela_de_desmatamento_ucs_atualizado_prodes-2023(1).xlsx', skiprows= 1, skipfooter=1, sheet_name = 'Cerrado')
df_desmatCe.drop(['TOTAL', 'Até 2000', 2002,2004,2006], axis=1, inplace=True)
df_desmatCa = pd.read_excel('https://github.com/JuPLopez/JL_mba_enap/raw/refs/heads/main/Estatistica_Descritiva/tabela_de_desmatamento_ucs_atualizado_prodes-2023(1).xlsx', skiprows= 1, skipfooter=1, sheet_name = 'Caatinga')
df_desmatCa.drop(['TOTAL', 'Até 2000', '2001 a 2004',2006], axis=1, inplace=True)

```

### Exercício 4. Exiba a distribuição de frequência de cada variável

```python
# (2) área_ha_uc: variável quantivativa contínua
df_baseED['área_ha_uc'].describe()
```

```python
def area(x):
  try:
      x = int(x)
  except:
      return pd.NA

  if x<1000: return "<1000"
  elif x<10000: return "1000-9999"
  elif x<100000: return "10000-99999"
  elif x>100000: return "100000-999999"
  elif x>1000000: return ">=1000000"

  return pd.NA

df_baseED['area'] = df_baseED['área_ha_uc'].apply(area);
print(df_baseED.area.value_counts().sort_index())
df_baseED.area.hist()
```

```python
# (3) Até 2007, (4) 2008, (5) 2009, (6) 2010, (7) 2011, (8) 2012, (9) 2013, (10) 2014, (11) 2015, (12) 2016, (13) 2017, (14) 2018, (15) 2019, (16) 2020, (17) 2021, (18) 2022, (19) 2023: variáveis quantivativas contínuas
df_baseED[2012].describe()
```

```python
df_baseED[2012].hist()
```

```python
df_baseED[2023].describe()
```

```python
# (20) UF: variável qualitativa nominal
#df_baseED.UF.value_counts() 
```

```python
# (22) Plano de Manejo: variável qualitativa nominal
df_baseED['Plano de Manejo'].value_counts() 
```

```python
# (23) Categoria (Sigla): variável qualitativa nominal
df_baseED['Categoria (Sigla)'].value_counts() 
df_baseED['Categoria (Sigla)'].hist()
```

```python
# (24) Grupo (Sigla): variável qualitativa nominal
df_baseED['Grupo (Sigla)'].value_counts()
df_baseED['Grupo (Sigla)'].hist()
```

```python
# (25) Bioma: variável qualitativa nominal
df_baseED['Bioma'].value_counts()
df_baseED['Bioma'].hist()
```

```python
# (26) Ano de Criação: variável quantitativa discreta
def criacao(x):
  try:
      x = int(x)
  except:
      return pd.NA

  if x<1960: return "<1960"
  elif x<1970: return "1960-1969"
  elif x<1980: return "1970-1979"
  elif x<1990: return "1980-1989"
  elif x<2000: return "1990-1999"
  elif x<2010: return "2000-2009"
  elif x<2020: return "2010-2019"
  elif x>=2020: return ">2020"

  return pd.NA

df_baseED['faixas'] = df_baseED['Ano de Criação'].apply(criacao);
print(df_baseED.faixas.value_counts().sort_index())
df_baseED.faixas.hist()
```
